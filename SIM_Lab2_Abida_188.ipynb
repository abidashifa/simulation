```python
import numpy as np

d1 = 8
d2 = 8
A = np.array([
    [d1 + 2, d2 + 1],
    [2 * d1, d2 + 2]
])
print("d1 =", d1)
print("d2 =", d2)
print("Matrix A:")
print(A)
```

```python
import numpy as np
shape_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues_A = np.linalg.eigvals(A)

print("1. Shape of A:", shape_A)
print("2. Determinant of A:", det_A)
print("3. Rank of A:", rank_A)
print("4. Eigenvalues of A:", eigenvalues_A)

if not np.isclose(det_A, 0):
    inv_A = np.linalg.inv(A)
    print("5. Inverse of A:\n", inv_A)
else:
    print("5. Inverse of A: Does not exist (Determinant is 0)")
```

```python
import numpy as np
B = A.copy()
B[0, 0] = B[0, 0] + 1

print("We changed element A[0,0] from 10 to:", B[0, 0])
print("\nMatrix B:")
print(B)
```
