```python
matrix_2x3 = np.array([[2.1, 4.7, -1.3],
                       [5.5, 0.2, 3.9]])

print("Original 2x3 Matrix:")
print(matrix_2x3)

min_val = np.min(matrix_2x3)
max_val = np.max(matrix_2x3)
min_idx = np.unravel_index(np.argmin(matrix_2x3), matrix_2x3.shape)
max_idx = np.unravel_index(np.argmax(matrix_2x3), matrix_2x3.shape)

print(f"\nMin Value: {min_val} at index {min_idx}")
print(f"Max Value: {max_val} at index {max_idx}")

vector_length = np.linalg.norm(matrix_2x3)
print(f"\nVector Length (L2 Norm of matrix): {vector_length:.4f}")

sorted_row_desc = -np.sort(-matrix_2x3, axis=1)
sorted_entire_desc = -np.sort(-matrix_2x3.flatten())

print("\nSorted Row-wise Descending:")
print(sorted_row_desc)
print("Sorted Entire Matrix Descending:")
print(sorted_entire_desc)

print(f"\nSum: {np.sum(matrix_2x3)}")
print(f"Product: {np.prod(matrix_2x3)}")
print(f"Median: {np.median(matrix_2x3)}")
print(f"Mean: {np.mean(matrix_2x3)}")
print(f"Standard Deviation: {np.std(matrix_2x3):.4f}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
np.random.seed(42)

random_matrix = np.random.uniform(low=-5.0, high=5.0, size=(4, 4))
print("Randomly Generated 4x4 Matrix:")
print(random_matrix)

plt.figure(figsize=(6, 5))
sns.heatmap(random_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Heatmap of Randomly Generated 4x4 Matrix")
plt.show()
```
