```python
import numpy as np
import scipy.stats as stats

data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])
```

```python
print("1. Hypotheses:")
print("   H0: mu1 = mu2 (The population means do not differ)")
print("   H1: mu1 != mu2 (The population means differ)")
print()
```

```python
mean1 = np.mean(data1)
mean2 = np.mean(data2)
var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)
mean_diff = mean1 - mean2

print(f"   Data 1 Mean: {mean1:.4f}, Variance (ddof=1): {var1:.4f}")
print(f"   Data 2 Mean: {mean2:.4f}, Variance (ddof=1): {var2:.4f}")
print(f"   Mean Difference (Data 1 - Data 2): {mean_diff:.4f}")
print()
```

```python
t_stat, p_val = stats.ttest_ind(data1, data2, equal_var=True)
n1, n2 = len(data1), len(data2)
df = n1 + n2 - 2

print(f"   t-statistic: {t_stat:.4f}")
print(f"   Degrees of freedom: {df}")
print(f"   Two-tailed p-value: {p_val:.4e}")
print()
```

```python
s_pooled = np.sqrt(((n1 - 1) * var1 + (n2 - 1) * var2) / df)
se_diff = s_pooled * np.sqrt(1/n1 + 1/n2)
t_critical = stats.t.ppf(0.975, df)

ci_lower = mean_diff - t_critical * se_diff
ci_upper = mean_diff + t_critical * se_diff

print("4. 95% Confidence Interval:")
print(f"   [{ci_lower:.4f}, {ci_upper:.4f}]")
print()
```

```python
alpha = 0.05
print("5. Decision & Interpretation:")
if p_val < alpha:
    print(f"   Since p-value ({p_val:.4e}) < {alpha}, we REJECT the null hypothesis (H0).")
    print("   Interpretation: There is a statistically significant difference between the population means of the two groups.")
else:
    print(f"   Since p-value ({p_val:.4e}) >= {alpha}, we FAIL TO REJECT the null hypothesis (H0).")
    print("   Interpretation: There is not enough statistical evidence to conclude that the population means of the two groups differ.")
```
